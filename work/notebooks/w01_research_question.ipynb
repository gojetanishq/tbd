## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

total_pages = len(df)
declining_pages = (df["trend_direction"] == "down").sum()
declining_rate = declining_pages / total_pages
average_impressions = df["impressions_90d"].mean()
median_word_count = df["word_count"].median()

print(f"Total pages: {total_pages}")
print(f"Declining pages: {declining_pages}")
print(f"Declining rate: {declining_rate:.3f}")
print(f"Average impressions: {average_impressions:.2f}")
print(f"Median word count: {median_word_count:.0f}")
```
